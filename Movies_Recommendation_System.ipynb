# **NLP**

```python
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import re
```

```python
# download stopwords and wordnet
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
stop_words = set(stopwords.words('english')) # set stopwords using variable
lemmatizer = WordNetLemmatizer()  # set lemmatizer using variable
```

```python
def preprocess_text(text):
  # convert text into lower case
  text = str(text).lower()

  # remove all punctuation
  text = re.sub(r'[^\w\s]', '', text)

  words = text.split()
  # remove stopwords
  words = [word for word in words if word not in stop_words]

  # lemmatize the words
  words = [lemmatizer.lemmatize(word) for word in words]

  return ' '.join(words)
```

```python
# now apply this function on tags
df['tags'] = df['tags'].apply(preprocess_text)
```

```python
df['tags'][1] # you can see all stopwords are removed
```

```python
# reset all database
df = df.reset_index(drop=True)
```

```python
indices = pd.Series(df.index, index=df['title']).drop_duplicates()
indices
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf = TfidfVectorizer(max_features=50000,ngram_range=(1,2),stop_words='english')
```

```python
tfidf_matrix = tfidf.fit_transform(df['tags'])
```

```python
tfidf_matrix
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
def recommend(title, n = 10):
  if title not in indices:
    return ['Movie not found']

  idx = indices[title]
  sim_score = cosine_similarity(tfidf_matrix[idx], tfidf_matrix).flatten()
  similar_idx = sim_score.argsort()[::-1][1:n+1]
  return df['title'].iloc[similar_idx]
```

```python
recommend('Avatar')
```

```python
import pickle

pickle.dump(tfidf_matrix,open('tfidf_matrix.pkl','wb'))

pickle.dump(indices,open('indices.pkl','wb'))

df.to_pickle('df.pkl')

pickle.dump(tfidf,open('tfidf.pkl','wb'))
```
